```python
import pandas as pd
```

```python
customers = pd.read_csv("dataset\olist_customers_dataset.csv")
print(customers.head())
```

```python
valid_records = []
rejected_records = []

for index, row in customers.iterrows():

    if row.isnull().any():
        rejected_records.append(row)
    else:
        valid_records.append(row)

print("Valid records:", len(valid_records))
print("Rejected records:", len(rejected_records))
```

3

```python
bad_records = pd.DataFrame(rejected_records)

bad_records.to_csv("bad_records.csv", index=False)

print("Rejected records saved successfully")
```

```python
order_items = pd.read_csv("dataset/olist_order_items_dataset.csv")

order_items["price"] = pd.to_numeric(
    order_items["price"], errors="coerce"
)

print("Invalid prices:", order_items["price"].isnull().sum())
```

```python
orders = pd.read_csv("dataset/olist_orders_dataset.csv")

orders["order_purchase_timestamp"] = pd.to_datetime(
    orders["order_purchase_timestamp"], errors="coerce"
)

print("Invalid dates:", orders["order_purchase_timestamp"].isnull().sum())
```

```python
invalid_price = order_items[order_items["price"] <= 0]

print("Invalid prices:", len(invalid_price))
```

```python
allowed_status = [
    "created",
    "approved",
    "invoiced",
    "processing",
    "shipped",
    "delivered",
    "unavailable",
    "canceled"
]
```

```python
invalid_status = orders[~orders["order_status"].isin(allowed_status)]

print("Invalid order status:", len(invalid_status))
```
